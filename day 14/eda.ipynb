```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed", "Jasim", "Ali"],
    "Age": [20, 22, 21, np.nan, 20],
    "Marks": [78, 92, 85, 67, 78],
    "Department": ["CS", "IT", "CS", "SE", "CS"]
})

print(df.head())

print(df.shape)

print(df.describe())

print(df.isnull().sum())

print(df.duplicated().sum())

print(df["Department"].unique())

print(df["Department"].value_counts())

print(df[df["Marks"] > 80])

print(df.groupby("Department")["Marks"].mean())

print(df["Marks"].mean())

print(df["Marks"].max())

print(df["Marks"].min())

print(df.corr(numeric_only=True))

plt.hist(df["Marks"], bins=5)
plt.show()

plt.scatter(df["Age"], df["Marks"])
plt.show()

plt.bar(df["Department"].value_counts().index,
        df["Department"].value_counts().values)
plt.show()
```
