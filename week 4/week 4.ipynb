```python
import matplotlib.pyplot as plt
import numpy as np

# Sample Data
movies = ["A", "B", "C", "D", "E"]
revenue = [120, 150, 180, 170, 200]
runtime = [90, 110, 130, 120, 140]
votes = [1000, 2500, 3000, 2800, 4000]
metascore = [70, 80, 75, 85, 90]

# Line Plot
plt.figure(figsize=(5,3))
plt.plot(movies, revenue)
plt.title("Movie Revenue")
plt.show()

# IMDb Movie Revenue Line Plot
plt.figure(figsize=(5,3))
plt.plot(movies, revenue, marker="o")
plt.xlabel("Movies")
plt.ylabel("Revenue")
plt.title("IMDb Movie Revenue")
plt.show()

# Time vs Runtime
plt.figure(figsize=(5,3))
plt.plot(movies, runtime)
plt.xlabel("Movies")
plt.ylabel("Runtime")
plt.title("Runtime")
plt.show()

# Votes vs Metascore
plt.figure(figsize=(5,3))
plt.plot(votes, metascore)
plt.xlabel("Votes")
plt.ylabel("Metascore")
plt.title("Votes vs Metascore")
plt.show()

# Line Styling and Labels
plt.figure(figsize=(5,3))
plt.plot(movies, revenue, color="red", linestyle="--", marker="o", linewidth=2)
plt.xlabel("Movies")
plt.ylabel("Revenue")
plt.title("Styled Line Plot")
plt.grid(True)
plt.show()

# Scatter Plot
plt.figure(figsize=(5,3))
plt.scatter(runtime, revenue)
plt.xlabel("Runtime")
plt.ylabel("Revenue")
plt.title("Scatter Plot")
plt.show()

# Bar Plot
plt.figure(figsize=(5,3))
plt.bar(movies, revenue)
plt.title("Revenue Bar Chart")
plt.show()

# Histogram
plt.figure(figsize=(5,3))
plt.hist(runtime, bins=5)
plt.title("Runtime Distribution")
plt.show()

# Subplots
fig, ax = plt.subplots(1, 2, figsize=(8,3))
ax[0].plot(movies, revenue)
ax[0].set_title("Revenue")

ax[1].bar(movies, revenue)
ax[1].set_title("Bar Chart")

plt.tight_layout()
plt.show()

# Zoomed Figure
plt.figure(figsize=(5,3))
plt.plot(votes, revenue, marker="o")
plt.xlim(2000, 4500)
plt.ylim(140, 210)
plt.title("Zoomed Plot")
plt.show()

# Pie Chart
plt.figure(figsize=(5,3))
plt.pie(revenue, labels=movies, autopct="%1.1f%%")
plt.title("Revenue Share")
plt.show()

# Figure Size Example
plt.figure(figsize=(8,4))
plt.plot(runtime, votes)
plt.xlabel("Runtime")
plt.ylabel("Votes")
plt.title("Custom Figure Size")
plt.show()
```
