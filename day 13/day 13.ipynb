```python
import pandas as pd
import numpy as np

# DataFrame Creation
df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed"],
    "Age": [20, 21, 22],
    "Marks": [85, 90, 78]
})
print(df)

# Read CSV File
# df = pd.read_csv("students.csv")

# Column Manipulation
df["City"] = ["Lahore", "Islamabad", "Karachi"]
print(df)

# Arithmetic Operations
df["Marks"] = df["Marks"] + 5
print(df)

# Null Value Handling
df.loc[1, "Age"] = np.nan
print(df.isnull())
df["Age"] = df["Age"].fillna(df["Age"].mean())
print(df)

# Data Filtering
print(df[df["Marks"] > 85])

# Unique Values
print(df["City"].unique())

# Duplicate Values
print(df.duplicated())

# Retrieve Rows
print(df.loc[0])
print(df.iloc[1])

# Replace Values
df["City"] = df["City"].replace("Lahore", "Rawalpindi")
print(df)

# Rename Columns
df.rename(columns={"Marks": "Score"}, inplace=True)
print(df)

# Lambda Apply
df["Result"] = df["Score"].apply(lambda x: "Pass" if x >= 85 else "Fail")
print(df)

# GroupBy
print(df.groupby("Result")["Score"].mean())

# GroupBy Multiple Columns
df["Department"] = ["IT", "HR", "IT"]
print(df.groupby(["Department", "Result"])["Score"].mean())

# Concatenation
df2 = pd.DataFrame({
    "Name": ["Hamza"],
    "Age": [23],
    "Score": [88],
    "City": ["Multan"],
    "Result": ["Pass"],
    "Department": ["HR"]
})
print(pd.concat([df, df2], ignore_index=True))

# Merge
student = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Ahmed"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Score": [80, 90, 85]
})

print(pd.merge(student, marks, on="ID"))

# Join
dept = pd.DataFrame({"City": ["Lahore", "Islamabad", "Karachi"]})
print(student.join(dept))

# Date and Time
dates = pd.DataFrame({
    "Date": ["2026-01-01", "2026-02-15", "2026-03-10"]
})

dates["Date"] = pd.to_datetime(dates["Date"])
print(dates["Date"].dt.year)
print(dates["Date"].dt.month)
print(dates["Date"].dt.day)
```
