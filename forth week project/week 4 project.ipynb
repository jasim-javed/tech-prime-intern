```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Title": ["Stranger Things","Wednesday","Money Heist","Dark","Squid Game","The Witcher","Lupin","Narcos","Breaking Bad","Lucifer"],
    "Genre": ["Sci-Fi","Horror","Crime","Sci-Fi","Thriller","Fantasy","Crime","Crime","Drama","Fantasy"],
    "Year": [2016,2022,2017,2017,2021,2019,2021,2015,2008,2016],
    "Rating": [8.7,8.1,8.2,8.8,8.0,8.0,7.5,8.8,9.5,8.1],
    "Seasons": [4,1,5,3,1,3,2,3,5,6],
    "Views(M)": [140,120,110,85,150,90,70,80,160,95]
}

df = pd.DataFrame(data)

print("Netflix Dataset")
print(df)

print("\nFirst 5 Rows")
print(df.head())

print("\nDataset Information")
print(df.info())

print("\nStatistical Summary")
print(df.describe())

print("\nHighest Rated Show")
print(df[df["Rating"] == df["Rating"].max()])

print("\nLowest Rated Show")
print(df[df["Rating"] == df["Rating"].min()])

print("\nShows Released After 2018")
print(df[df["Year"] > 2018])

print("\nCrime Shows")
print(df[df["Genre"] == "Crime"])

print("\nAverage Rating:", df["Rating"].mean())
print("Average Views:", df["Views(M)"].mean())

print("\nGenre Counts")
print(df["Genre"].value_counts())

print("\nTop 3 Most Viewed Shows")
print(df.nlargest(3, "Views(M)"))

df["Popularity"] = df["Views(M)"].apply(lambda x: "High" if x >= 120 else "Medium" if x >= 90 else "Low")
print("\nPopularity")
print(df[["Title","Popularity"]])

print("\nAverage Views by Genre")
print(df.groupby("Genre")["Views(M)"].mean())

plt.figure(figsize=(6,4))
plt.bar(df["Title"], df["Views(M)"])
plt.xticks(rotation=45)
plt.title("Netflix Show Views")
plt.xlabel("Shows")
plt.ylabel("Views (Millions)")
plt.tight_layout()
plt.show()

plt.figure(figsize=(6,4))
plt.scatter(df["Rating"], df["Views(M)"])
plt.xlabel("Rating")
plt.ylabel("Views (Millions)")
plt.title("Rating vs Views")
plt.show()

plt.figure(figsize=(6,4))
plt.plot(df["Year"], df["Views(M)"], marker="o")
plt.xlabel("Year")
plt.ylabel("Views (Millions)")
plt.title("Views by Release Year")
plt.grid(True)
plt.show()

plt.figure(figsize=(5,5))
plt.pie(df["Genre"].value_counts(),
        labels=df["Genre"].value_counts().index,
        autopct="%1.1f%%")
plt.title("Genre Distribution")
plt.show()
```
