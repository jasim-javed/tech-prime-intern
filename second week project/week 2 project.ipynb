```python
import numpy as np
```

```python
# Patient IDs
patient_ids = np.arange(1,11)

# Patient Ages
ages = np.array([25,42,36,50,61,29,40,55,48,33])

# Heart Rate
heart_rate = np.array([72,88,76,90,85,70,79,95,82,74])

# Temperature
temperature = np.array([98.6,99.2,98.4,100.1,99.5,98.2,98.8,101.0,99.0,98.5])

print(patient_ids)
```

```python
print("Patient IDs")
print(patient_ids)

print("\nPatient Ages")
print(ages)

print("\nHeart Rate")
print(heart_rate)

print("\nTemperature")
print(temperature)
```

```python
print("Shape of Age Array :",ages.shape)
print("Shape of Heart Rate :",heart_rate.shape)
```

```python
print("Average Age :",np.mean(ages))
print("Minimum Age :",np.min(ages))
print("Maximum Age :",np.max(ages))

print("Average Heart Rate :",np.mean(heart_rate))

print("Highest Temperature :",np.max(temperature))
print("Lowest Temperature :",np.min(temperature))
```

```python
print("Total Age :",np.sum(ages))
print("Total Heart Rate :",np.sum(heart_rate))
```

```python
print("First Patient Age :",ages[0])

print("Last Patient Age :",ages[-1])

print("Third Patient Temperature :",temperature[2])
```

```python
print("First Five Ages")

print(ages[:5])

print("Last Three Heart Rates")

print(heart_rate[-3:])
```

```python
fever = np.where(temperature>99)

print(fever)

print("Patients having fever")

print(patient_ids[fever])
```

```python
old_patients=np.where(ages>40)

print(patient_ids[old_patients])
```

```python
high=np.where(heart_rate>85)

print(patient_ids[high])
```

```python
patient_matrix=ages.reshape(2,5)

print(patient_matrix)
```

```python
print("Hospital Report ")

print("Total Patients :",len(patient_ids))

print("Average Age :",np.mean(ages))

print("Average Heart Rate :",np.mean(heart_rate))

print("Highest Temperature :",np.max(temperature))

print("Patients with Fever :",patient_ids[np.where(temperature>99)])

print("Patients Above 40 :",patient_ids[np.where(ages>40)])
```
